Initialize two different types of Components (exponential and weibull)

```python
from shipClass_continous.component_objects.component_exponential import ExponentialComponent
from shipClass_continous.component_objects.component_weibull import WeibullComponent

# import logging
import matplotlib.pyplot as plt
import numpy as np

# set logging level here
# logging.basicConfig(level=logging.WARNING)

# simulate an exponential component
pump = ExponentialComponent(name="Pump A", MTTF=100, MTTR=10)
pump.simulate(t_end=500, dt=1)  # 500 time units, 1 per step
print(pump)

# simulate a weibull component
pump2 = WeibullComponent(name="Pump B", MTTF=100, MTTR=10, shape=1.5)
pump2.simulate(t_end=500, dt=1)  # 500 time units, 1 per step
print(pump2)

# plot the histories of both components in a single plot
print(pump.history.shape)

plt.figure(figsize=(10, 5))
plt.plot(pump.history[:,0], pump.history[:,1], label="Pump A (Exponential)")
plt.plot(pump2.history[:,0], pump2.history[:,1], label="Pump B (Weibull)")
plt.xlabel("Time")
plt.ylabel("State")
plt.title("Component State Over Time")
plt.legend()
plt.grid()
plt.show()
```
