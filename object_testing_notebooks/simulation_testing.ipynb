Checking Initalization of the TA-O systems

```python
from simulations.initialization import build_TAO
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (7, 4) # width, height

# initialize and simulate TAO system
TAO_system = build_TAO()
TAO_system.simulate(7500)

# Plot histories of individual subsystems
fig, axs = plt.subplots(
    nrows=3,
    ncols=1,
    figsize=(10, 8),
    sharex=True
    )
TAO_system.components[0].plot_history(ax=axs[0], plot_comps=True)  # fuel_oil_sys
TAO_system.components[2].plot_history(ax=axs[1], plot_comps=True)  # transmission_sys
TAO_system.components[1].plot_history(ax=axs[2], plot_comps=True)  # parallel_engines
plt.tight_layout()
plt.show()

# Plot history of the entire TAO system
TAO_system.plot_history()
```
